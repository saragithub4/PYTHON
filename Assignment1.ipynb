String concatenation

```python
name=input("Enter your name:")
str1="Hello "
str2=name
result=str1+ str2
print(result)
```

```python
str3=", welcome to Python Programming"
result=result+str3
print(result)
```

String Slicing and Indexing

```python
#Print the first character of the string
string=result
print("First character:", string[0])

#Print the last character of the string
print("Last character:",string[-1])

#Print the first 5 characters of the string
print("First 5 characters:",string[:5])

#Print the last 11 characters of the string
print("Last 11 characters:",string[-11:])

#Print the string in reverse
print("Reverse:",string[::-1])

#Use slicing and print the word “Python” from the existing string
start=string.index("Python")
print("Word Python:",string[start:start+6])
```

String Methods

```python
strM="Python beginner tutorial"
print(strM)

#Convert the sentence to uppercase
print("Uppercase:",strM.upper())

#Convert the sentence to lowercase
print("Lower case:",strM.lower())

#Use Capitalize and return the sentence to the original input form
print("Capitalised:",strM.capitalize())

#Count the total number of occurrences of character ‘t’ in the string
print("Number of 't' occurrences:",strM.count("t"))

#Replace all occurrences of “Python” with “Machine Learning"
print("Replaced:",strM.replace("Python","Machine Learning"))
```

Tuples

```python
t1=(10,20,30)
t2=(40,50,60)

#Concatenate the two tuples and store it in “t_combine”
t_combine=t1+t2
print(t_combine)

#Repeat the elements of “t_combine” 3 times
print("Repeated 3 times:",t_combine*3)

#Access the 3rd element from “t_combine”
print("3rd Element:",t_combine[2])

#Access the first three elements from “t_combine”
print("First 3 elements:",t_combine[:3])

#Access the last three elements from “t_combine”
print("Last 3 elements:",t_combine[-3:])
```
